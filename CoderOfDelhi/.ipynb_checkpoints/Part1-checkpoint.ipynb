```python
import json
```

```python
def loadData(filename):
    with open(filename, "r") as file:
        data = json.load(file)
    return data
```

```python
data = loadData("data.json")
data
```

```python
def displayUser(data):

    #Basically created a quick look up for fast access of data userwise when required
    idToName = {u['id']:u['name'] for u in data['users']}
    pageName = {p['id']:p['name'] for p in data['pages']}
    
    for user in data['users']:
        # this is taking the id from frieds arr above and then taking name from the lookup table and adding it to friendName array
        # similar thing is done by pageName array also as id of pages is present as liked_pages
        friendName = [idToName[fID] for fID in user['friends']]
        pageNames = [pageName[pid] for pid in user['liked_pages']]
        print(f"{user['name']} is friends with {', '.join(friendName)} and liked {', '.join(pageNames)}")

    print("\n")
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")
```

```python
displayUser(data)
```
