```python
import json
```

```python
def loadData(filename):
    with open(filename, "r") as f:
        return json.load(f)
```

```python
data = loadData("data3.json")
```

```python
def pagesYouMightLike(userId, data):
    userPages = {}
    for user in data['users']:
        userPages[user['id']] = set(user['liked_pages'])

    if userId not in userPages:
        return []

    userLikedPages = userPages[userId]
    pagesToSuggest = {}

    for otherUsers, pages in userPages.items():
        if otherUsers != userId:
            sharedPages = userLikedPages.intersection(pages)
            for page in pages:
                if page not in userLikedPages:
                    pagesToSuggest[page] = pagesToSuggest.get(page, 0) + len(sharedPages)

    sorted_suggestions = sorted(pagesToSuggest.items(), key = lambda x:x[1], reverse = True)
    return [(pageId, score) for pageId, score in sorted_suggestions if score >= 1]
```

```python
userId = 1
txt = pagesYouMightLike(userId,data)
txt
```

```python
pageNameMapping = {p['id']: p['name'] for p in data['pages']}
pageNameMapping
```

```python
print(f"You might also like these articles: {data['users']['id']['name']}")
for id, (page, _) in enumerate(txt, start = 1):
    print(f"{id}. {pageNameMapping[page]}")
```
