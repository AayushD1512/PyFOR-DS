```python
import json

def loadData(filename):
    with open(filename, 'r') as file:
        data = json.load(file)
    return data
```

```python
data = loadData("data.json")
data
```

```python
def displayUser(data):

    friendsRelation = {u['id']:u['name'] for u in data['users']}
    pageNames = {p['id']:p['name'] for p in data['pages']}
    
    for user in data['users']:
        friendName = [friendsRelation[id] for id in user['friends']]
        pageName = [pageNames[id] for id in user['liked_pages']]
        
        print(f"{user['id']}. {user['name']} is friends with {','.join(friendName)} and likes {','.join(pageName)}")
```

```python
displayUser(data)
```
