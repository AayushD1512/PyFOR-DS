```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
chunks[3]

def chunkParser(chunks):
    chunk = chunks.split("\n")
    username = chunk[0].strip()
    post = chunk[1].replace("posts", "").replace(",", "").strip()
    followers = chunk[2].replace("followers","").strip().replace(".","").replace("K","000")
    following = chunk[3].replace("following","").strip()
    name = chunk[4].split(" ")[0]
    bio = ",".join(chunk[5:])
    

    return {
        "username": username,
        "post": post,
        "followers": followers,
        "following": following,
        "name": name,
        "bio": bio}

test = chunkParser(chunks[3])
test
```
