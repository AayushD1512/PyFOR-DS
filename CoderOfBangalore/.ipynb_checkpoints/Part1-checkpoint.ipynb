```python
with open("final.txt", encoding="utf-8") as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]

def chunkParser(chunks):
    try:
        chunks = chunks.strip()
        chunk = chunks.split("\n")
        username = chunk[0].strip()
        post = int(chunk[1].replace("posts", "").replace(",", "").replace("post","").strip())
        
        followers = float(chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M",""))
        if ("K" in chunk[2]):
            followers = int(followers * 1000)
        elif ("M" in chunk[2]):
            followers = int(followers * 1000000)
        else:
            followers = int(followers)
        
        
        following = float(chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M",""))
        if ("K" in chunk[3]):
            following = int(following * 1000)
        elif ("M" in chunk[3]):
            following = int(following * 1000000)
        else:
            following = int(following)
            
        name = chunk[4].split(" ")[0]

        if len(chunk)>5:
            category = chunk[5].strip()
        else:
            category = "Unknown"
        if not category:
            category = "Unknown"
        bio = ",".join(chunk[6:])
    

        return {
            "username": username,
            "post": post,
            "followers": followers,
            "following": following,
            "name": name,
            "category": category,
            "bio": bio
        }

    except ValueError as e:
        print(e, chunk)
```

```python

allChunks = [chunkParser(chunk) for chunk in chunks]


import json
s = json.dumps(allChunks, indent=4)
with open("finalData.json", "w") as f:
    f.write(s)
```
