```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
year = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
yuvraj = [300.0, 600.0, 800.0, 1100.0, 900.0, 0, 0, 0, 0, 0]
```

```python
plt.bar(year, sachin)
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Sachin's Run over the years")
```

```python
x = np.arange(len(year))
width = 0.25

plt.bar(x-width, sachin, width=width, color="red", label="Sachin")
plt.bar(x, kohli, width=width, color="green", label="Kohli")
plt.bar(x+width, sehwag, width=width, color="blue", label="Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Runs Comparison over the years")
plt.xticks(x, year)
plt.legend()
plt.tight_layout()
```

```python
x = np.arange(len(year))
width = 0.45

plt.bar(x-width, sachin, width=width, color="red", label="Sachin")
plt.bar(x, kohli, width=width, color="green", label="Kohli")
plt.bar(x+width, sehwag, width=width, color="blue", label="Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Runs Comparison over the years")
plt.xticks(x, year)
plt.legend()
plt.tight_layout()
```

```python
sachinTotal = sum(sachin[:5])
kohliTotal = sum(kohli[:5])
sehwagTotal = sum(sehwag[:5])
yuvrajTotal = sum(yuvraj[:5])

playersSum = [sachinTotal, kohliTotal, yuvrajTotal, sehwagTotal]
names = ["sachin", "kohli", "yuvraj", "sehwag"]
```

```python
plt.barh(names, [sachinTotal, kohliTotal, yuvrajTotal, sehwagTotal])

plt.xlabel("Total run in 5 years")
plt.ylabel("Players")
plt.title("First 5 yr comparison")

for i in range(4):
    
    plt.text(playersSum[i], i, str(playersSum[i]), color="red", va="center")
```

```python
maxOfSachin = max(sachin)
maxOfKohli = max(kohli)
maxOfSehwag = max(sehwag)
maxOfYuvraj = max(yuvraj)

players = ["Sachin", "Kohli", "Sehwag", "Yuvraj"]

```

```python
maxRuns = [maxOfSachin, maxOfKohli, maxOfSehwag, maxOfYuvraj]
```

```python
plt.bar(players, maxRuns)

for i in range(len(players)):
    plt.text(
        i,
        maxRuns[i] + 50,
        str(maxRuns[i]),
        color="magenta",
        ha="center"
        
    )

plt.xlabel("Players")
plt.ylabel("Max Runs")
```
