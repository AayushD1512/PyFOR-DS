```python
import matplotlib.pyplot as plt
```

```python
year = [2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020]

messi = [59, 91, 45, 58, 52, 59, 54, 51, 50, 27]
ronaldo = [60, 63, 69, 61, 57, 55, 53, 49, 39, 44]
neymar = [40, 56, 37, 35, 46, 31, 34, 34, 12, 11]
```

```python
plt.style.use("ggplot")
plt.plot(year, messi, label='Messi', color="red", linestyle=':')
plt.plot(year, ronaldo, label='Ronaldo', color="black", linestyle='dotted')
plt.plot(year, neymar, label='Neymar', color="purple", linestyle='dashdot')
```
