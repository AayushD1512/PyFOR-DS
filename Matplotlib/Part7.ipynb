```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y1 = [i * 2 for i in x]
y2 = [i ** 2 for i in x]
y3 = [i ** 0.5 for i in x]
y4 = [10 - i for i in x]
```

```python
fig, axs = plt.subplots(2,2, figsize = (10,4))
axs = axs.flatten()

axs[0].plot(x, y1)
axs[0].set_title("x*2")

axs[1].plot(x, y2)
axs[1].set_title("x**2")

axs[2].plot(x, y3)
axs[2].set_title("x root 2")

axs[3].plot(x, y4)
axs[3].set_title("10 - x")

fig.suptitle("Some comparisons")fig.tight_layout()
fig.subplots_adjust(top=0.85)
```
