```python
import matplotlib.pyplot as plt
```

```python
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
```

```python
# Simple Pie Chart
```

```python
plt.pie(runs, labels=labels, colors=colors)
```

```python
plt.pie(
    runs, 
    labels=labels, 
    colors=colors,
    wedgeprops={'edgecolor':'black', 'linewidth':2, 'linestyle':'--'}
)
```

```python
#3D area AND SHADOW

explode = [0.15, 0, 0, 0]

plt.pie(
    runs, 
    labels=labels, 
    colors=colors,
    wedgeprops={'edgecolor':'black', 'linewidth':2, 'linestyle':'-'},
    explode=explode,
    shadow=True
)
```

```python
#showing percentage

#3D area AND SHADOW

explode = [0.15, 0, 0, 0]

plt.pie(
    runs, 
    labels=labels, 
    colors=colors,
    wedgeprops={'edgecolor':'black', 'linewidth':2, 'linestyle':'-'},
    explode=explode,
    shadow=True,
    autopct='%1.2f%%'
)
```

```python
languages = ['Python', 'Java', 'C++', 'JavaScript', 'C#', 'Ruby', 'Go', 'Rust', 'Swift', 'PHP']
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]
```

```python
#trying above with bar charts h and v

plt.bar(languages, usage, label=languages)
```

```python
plt.barh(languages, usage, label=languages)

for i in range(len(languages)):
    plt.text(
        usage[i]+0.7, i, str(usage[i]), color='black', va='center'
    )
```
