```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
pd.merge(employees, departments, on="DeptID") #inner join all data avail in both tables
```

```python
pd.merge(employees, departments, on="DeptID", how="left") #left join - all data from left table should be present even if NaN
```

```python
pd.merge(employees, departments, on="DeptID", how="right") #right join - all data from right table should be present even if NaN
```

```python
pd.merge(employees, departments, on="DeptID", how="outer") #outer join - everything from both tables
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})

```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```
