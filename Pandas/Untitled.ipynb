```python
import pandas as pd
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 78, 92],
    'Science': [90, 82, 89],
    'English': [88, 85, 94]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df = df.melt(id_vars=["Name"], value_vars=["Math", "Science", "English"], var_name="Subject", value_name="Score")
```

```python
df
```

```python
df2 = df.copy()
```

```python
df2
```

```python
df2 = df2.pivot_table(index="Name", columns="Subject", values="Score", aggfunc="mean").reset_index()
```

```python
df2.columns.name = None
```
