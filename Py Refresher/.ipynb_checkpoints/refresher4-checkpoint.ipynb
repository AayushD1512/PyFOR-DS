```python
def httpCode(code):
    match code:
        case "200":
            return "OK"
        case "400":
            return "Bad request"
        case "404":
            return "Page Note Found"
        case _:
            return "Unknown error"
```

```python
httpCode(404)
```

```python
point = (0, 4)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"X-Axis at {x}")
    case (0, y):
        print(f"Y-Axis at {y}")
    case (x, y):
        print(f"Point at ({x}, {y})")
```
